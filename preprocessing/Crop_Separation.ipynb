```python
import pandas as pd

# Load the dataset
df = pd.read_csv('/content/crop_yield.csv')

# Calculate the total number of rows
total_rows = len(df)

# Calculate percentage for Wheat
wheat_count = df[df['Crop'] == 'Wheat'].shape[0]
wheat_percentage = (wheat_count / total_rows) * 100 if total_rows > 0 else 0

# Calculate percentage for Rice
rice_count = df[df['Crop'] == 'Rice'].shape[0]
rice_percentage = (rice_count / total_rows) * 100 if total_rows > 0 else 0

# Calculate percentage for Maize
maize_count = df[df['Crop'] == 'Maize'].shape[0]
maize_percentage = (maize_count / total_rows) * 100 if total_rows > 0 else 0

print(f"Percentage of Wheat data: {wheat_percentage:.2f}%")
print(f"Percentage of Rice data: {rice_percentage:.2f}%")
print(f"Percentage of Maize data: {maize_percentage:.2f}%")

# Find the crop type with the highest distribution
crop_distribution = df['Crop'].value_counts()
highest_distribution_crop = crop_distribution.idxmax()
highest_distribution_count = crop_distribution.max()

print(f"\nThe crop type with the highest distribution is '{highest_distribution_crop}' with {highest_distribution_count} entries.")
```

```python
print("\nTop 5 crop types with the highest distribution:")
display(crop_distribution.head(5))
```

```python
# Create a DataFrame for only Rice crop data
df_rice = df[df['Crop'] == 'Rice'].copy()

print("First 5 rows of Rice DataFrame:")
display(df_rice.head())
```

```python
# Create a DataFrame for only Maize crop data
df_maize = df[df['Crop'] == 'Maize'].copy()

print("\nFirst 5 rows of Maize DataFrame:")
display(df_maize.head())
```

```python
# Save df_rice to a CSV file
df_rice.to_csv('rice_data.csv', index=False)
print("Rice data saved to 'rice_data.csv'")
```

```python
# Save df_maize to a CSV file
df_maize.to_csv('maize_data.csv', index=False)
print("Maize data saved to 'maize_data.csv'")
```
