```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("/content/drive/MyDrive/Crop Yield Prediction Using Multimodal Data Fusion/Datasets/Custom_Crops_yield_Historical_Dataset.csv")
```

```python
print(df.info())
```

```python
df.isnull().sum()
```

```python
selected_features = [
    'Year',
    'State Name',
    'Dist Name',
    'Crop',
    'Area_ha',
    'N_req_kg_per_ha',
    'P_req_kg_per_ha',
    'K_req_kg_per_ha',
    'Temperature_C',
    'Humidity_%',
    'pH',
    'Rainfall_mm',
    'Wind_Speed_m_s',
    'Solar_Radiation_MJ_m2_day',
    'Yield_kg_per_ha' # Target variable
]

df_selected = df[selected_features].copy()

print("Information about the DataFrame with selected features:")
display(df_selected.info())
print("\nFirst 5 rows of the selected DataFrame:")
display(df_selected.head())
```

```python
import pandas as pd

# One-hot encode 'State Name' and 'Crop'
df_processed = pd.get_dummies(df_selected, columns=['Crop'], drop_first=True)

# Drop 'Dist Name'
df_processed = df_processed.drop(columns=['Dist Name'])

print("DataFrame information after one-hot encoding and dropping 'Dist Name' (without scaling):")
display(df_processed.info())
print("\nFirst 5 rows of the processed DataFrame:")
display(df_processed.head())
```

```python
# Define the path to save the cleaned dataset
output_path = "/content/drive/MyDrive/Crop Yield Prediction Using Multimodal Data Fusion/Datasets/Cleaned_Crops_yield_Historical_Dataset.csv"

# Save the DataFrame to a CSV file
df_processed.to_csv(output_path, index=False)

print(f"Cleaned dataset saved successfully to: {output_path}")

# You can load it back later using:
# df_cleaned = pd.read_csv(output_path)
```
